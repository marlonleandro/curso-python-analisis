## Estructura de control condicional

Una estructura condicional en Python es una forma de que un programa tome decisiones, es decir, que ejecute ciertas instrucciones solo si se cumple una condición determinada.

La estructura condicional más común en Python es `if`, y se puede complementar con `elif` (else if) y `else`.

### Sintaxis básica:

```
if condicion:
    # Bloque de código si la condición es verdadera
elif otra_condicion:
    # Bloque de código si la primera no se cumple, pero esta sí
else:
    # Bloque de código si ninguna condición se cumple
```

````python
# Evaluar si un alumno aprobó o desaprobó
# 1. Declarar las variables
nota = 19.0
# 2. Evaluar la nota
if nota >= 10.5:
    print("El alumno APROBÓ")
else:
    print("El alumno DESAPROBÓ")
````

````python
# Expresar el siguiente comentario en código python
# "Si gasto hasta $100, pago con dinero en efectivo. Sino, si gasto más de $100 pero menos de $300, pago con tarjeta de débito.
# Sino, pago con tarjeta de crédito."

# 1. Declarar variable
gasto = float(input("¿Cuánto gastaste? $:"))    #float => convierte a decimal

# 2. Evaluar el gasto
if gasto <= 100:
    print(f"Usted gastó ${gasto}, por tanto paga con dinero en efectivo.")
elif gasto > 100 and gasto < 300:
    print(f"Usted gastó ${gasto}, por tanto paga con tarjeta de débito.")
else:
    print(f"Usted gastó ${gasto}, por tanto paga con tarjeta de crédito.")
````

````python
# Realizar un algortimo en python que calcule las raices de una ecuación cuadrática
# 1. Declarar variables (coeficientes a, b y c)
a = 1
b = -3
c = 2
# 2. Calcular la discriminante
d = b**2 - 4*a*c
# 3. Evaluar la discriminante
if d > 0:
    # Existen 2 raices reales
    r1 = (-b + d**0.5) / (2*a)
    r2 = (-b - d**0.5) / (2*a)
    print(f"Las raices de la ecuación son: {r1} y {r2}")
elif d == 0:
    # Existe una sola raiz
    r = -b / (2*a)
    print(f"La única la raiz es: {r}")
else:
    # No hay raices
    print("No hay solución en los números reales.")
````

## Estructura de control bucle

Una estructura de bucle en Python es una forma de repetir un bloque de código varias veces, ya sea un número determinado de veces o mientras se cumpla una condición.

Python tiene dos tipos principales de bucles:

### 1. for: Repetición conocida o definida

Se usa para recorrer elementos de una lista, rango de números, cadenas, etc.

#### Ejemplo:

```
for i in range(5):
    print("Hola", i)
```

### 2. while: Repetición condicional

Se ejecuta mientras se cumpla una condición lógica.

#### Ejemplo:

```
contador = 1
while contador <= 5:
    print("Vuelo número", contador)
    contador += 1
```

````python
# Mientras que el año sea menor o igual de 2025, imprimir la frase "Informes del Año xxxx"
anio = 2020
while anio <= 2025:
    print(f"Informes del Año {anio}")
    anio += 1   # Incrementar el año en 1
````

````python
# Imprimir los valores de los elementos de una lista
frutas = ["Mazana", "Naranja", "Fresa", "Plátano", "Pera", "Uva"]
# Recorrer la lista de frutas
for fru in frutas:
    print(f"Fruta seleccionada: {fru}")
````

````python
# Imprimir los números del 1 al 10
for num in range(1, 11):
    print(f"Número: {num}")
````

````python
# Imprimir los números del 1 al 10 (usando while)
num = 1
while num <= 10:
    print(f"Número: {num}")
    num += 1
````

````python
# Calcular el promedio de las siguientes notas:
notas = [11, 17, 15, 16, 13, 18, 14, 11]
# 1. Obtener la cantidad de notas
cantidad = len(notas)   # len => función que cuenta elementos de una lista
# 2. Sumar todas las notas
suma = 0    # Iniciar la suma en 0
for n in notas:
    suma += n
# 3. Calcular el promedio
prom = suma / cantidad
# 4. Mostrar resultado
print(f"El promedio es: {round(prom, 2)}")  # round => redondear decimales
````
