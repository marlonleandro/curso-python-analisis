## Estructura de control condicional

Una estructura condicional en Python es una forma de que un programa tome decisiones, es decir, que ejecute ciertas instrucciones solo si se cumple una condición determinada.

La estructura condicional más común en Python es `if`, y se puede complementar con `elif` (else if) y `else`.

### Sintaxis básica:

```
if condicion:
    # Bloque de código si la condición es verdadera
elif otra_condicion:
    # Bloque de código si la primera no se cumple, pero esta sí
else:
    # Bloque de código si ninguna condición se cumple
```

````python
# Ejercicio 1: Evaluar si un alumno aprobó o desaprobó
# 1. Declarar las variables
nota = 9.5

# 2. Evaluar la nota
if nota >= 10.5:
    print("El alumno APROBÓ")
else:
    print("El alumno DESAPROBÓ")
````

````python
# Practica 1: Calcular el promedio de tres notas e indicar si el alumno aprobó o desaprobó
# - considerando que la nota mínima para aprobar es 11
# 1. Declarar y asignar variables
nota1 = 12
nota2 = 5
nota3 = 10

# 2. Calcular el promedio
promedio = (nota1 + nota2 + nota3) / 3

# 3. Evaluar si aprobó o desaprobó
if promedio >= 11:
    print(f"El alumno APROBÓ con un promedio de: {promedio:.2f}")
else:
    print(f"El alumno DESAPROBÓ con un promedio de: {promedio:.2f}")
````

````python
# Expresar el siguiente comentario en código python
# "Si gasto hasta $100, pago con dinero en efectivo. Sino, si gasto más de $100 pero menos de $300, pago con tarjeta de débito.
# Sino, pago con tarjeta de crédito."

# 1. Declarar variable
gasto = float(input("¿Cuánto gastaste? $:"))    #float => convierte a decimal

# 2. Evaluar el gasto
if gasto <= 100:
    print(f"Usted gastó ${gasto}, por tanto paga con dinero en EFECTIVO.")
elif gasto > 100 and gasto < 300:
    print(f"Usted gastó ${gasto}, por tanto paga con tarjeta de DÉBITO.")
else:
    print(f"Usted gastó ${gasto}, por tanto paga con tarjeta de CRÉDITO.")
````

````python
# Practica 3: Clasificar la nota de un alumno (de 0 a 20) en:
# - 18 a 20 → Excelente
# - 14 a 17 → Bueno
# - 11 a 13 → Regular
# - 0 a 10 → Malo

nota = int(input("Ingrese la nota del alumno (0-20): "))

# Evaluando la nota
if nota >= 18 and nota <= 20:
    print("La nota es Excelente")
elif nota >= 14 and nota <= 17:
    print("La nota es Buena")
elif nota >= 11 and nota <= 13:
    print("La nota es Regular")
elif nota >= 0 and nota <= 10:
    print("La nota es Mala")
else:
    print("La nota ingresada no es válida. Debe estar entre 0 y 20.")
````

````python
# Realizar un algortimo en python que calcule las raices de una ecuación cuadrática
# 1. Declarar variables (coeficientes a, b y c)
a = 1
b = 1
c = -25

# 2. Calcular la discriminante
d = b**2 - 4*a*c

# 3. Evaluar la discriminante
if d > 0:
    # Existen 2 raices reales
    r1 = (-b + d**0.5) / (2*a)
    r2 = (-b - d**0.5) / (2*a)
    print(f"Las raices de la ecuación son: {r1:.2f} y {r2:.2f}")
elif d == 0:
    # Existe una sola raiz
    r = -b / (2*a)
    print(f"La única la raiz es: {r:.2f}")
else:
    # No hay raices
    print("No hay solución en los números reales.")
````

````python
# Practica 4: Calcular el sueldo neto de un trabajador considerando:
# - Sueldo base
# - Asignacion familiar (150 adicional si tiene hijos)
# - Descuento por sistema de pensiones (AFP => 11% del sueldo base, ONP => 9% del sueldo base)
# - Descuento el impuesto a la renta (5% del sueldo base)

sueldo_base = float(input("Ingrese el sueldo base del trabajador: "))
cantidad_hijos = int(input("Ingrese la cantidad de hijos del trabajador: "))
tipo_pension = input("Ingrese el tipo de pensión (AFP/ONP): ").strip().upper()

# Calculando la asignación familiar
asignacion_familiar = 150 if cantidad_hijos > 0 else 0 #If implicito

# Calculando los descuentos
descuento_pension = 0
if tipo_pension == "AFP":
    descuento_pension = sueldo_base * 0.11
elif tipo_pension == "ONP":
    descuento_pension = sueldo_base * 0.09

# Calculando el descuento por impuesto a la renta
descuento_renta = sueldo_base * 0.05

# Calculando el sueldo neto
sueldo_neto = sueldo_base + asignacion_familiar - descuento_pension - descuento_renta
print("El sueldo neto del trabajador es:", sueldo_neto)

````

## Estructura de control bucle

Una estructura de bucle en Python es una forma de repetir un bloque de código varias veces, ya sea un número determinado de veces o mientras se cumpla una condición.

Python tiene dos tipos principales de bucles:

### 1. for: Repetición conocida o definida

Se usa para recorrer elementos de una lista, rango de números, cadenas, etc.

#### Ejemplo:

```
for i in range(5):
    print("Hola", i)
```

### 2. while: Repetición condicional

Se ejecuta mientras se cumpla una condición lógica.

#### Ejemplo:

```
contador = 1
while contador <= 5:
    print("Vuelo número", contador)
    contador += 1
```

````python
# Mientras que el año sea menor o igual de 2026, imprimir la frase "Informes del Año xxxx"
anio = 2020
while anio <= 2026:
    print(f"Informes del Año {anio}")
    anio += 1   # Incrementar el año en 1
````

````python
# Imprimir los valores de los elementos de una lista
frutas = ["Mazana", "Naranja", "Fresa", "Plátano", "Pera", "Uva"]
# Recorrer la lista de frutas
for fru in frutas:
    print(f"Fruta seleccionada: {fru}")
````

````python
# Imprimir los números del 1 al 10
for num in range(1, 11):
    print(f"Número: {num}")
````

````python
# Imprimir los números del 1 al 10 (usando while)
num = 1
while num <= 10:
    print(f"Número: {num}")
    num += 1    #Incrementar en 1
````

````python
# Calcular el promedio de las siguientes notas:
notas = [11, 17, 15, 16, 13, 18, 14, 11]

# 1. Obtener la cantidad de notas
cantidad = len(notas)   # len => función que cuenta elementos de una lista

# 2. Sumar todas las notas
suma = 0    # Iniciar la suma en 0
for n in notas:
    suma += n

# 3. Calcular el promedio
prom = suma / cantidad

# 4. Mostrar resultado
print(f"El promedio es: {round(prom, 2)}")  # round => redondear decimales
````
