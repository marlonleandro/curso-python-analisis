## Manejo de Listas en Python

- Las listas en Python, al igual que las cadenas, son datos ordenados. Una diferencia importante es que es un tipo de dato derivado, también llamado contenedor.
- Las listas contienen como elementos a enteros, a cadenas o incluso a otras listas. Una lista es una secuencia ordenada de datos.
- Las operaciones mas comunes son `append()` (agrega un elemento al final) y `pop()` (saca un elemento del final). 
- La longitud de una lista (o tamaño) está definida por la cantidad de elementos. Utilizamos la función `len()` para este propósito.
- Para acceder a los elementos de una lista lo hacemos mediante su índice que representa la posición del elemento empezando desde 0.

```python
# Listas de ejemplo
numeros = [10, 20, 30, 10]
nombres = ["Ana", "Luis", "Maria"]
mixta = ["Lima", 200, True, 125.88]

# Hallando la longitud de una lista
print(f"La longitud de la lista 'numeros' es: {len(numeros)}")

# Obtener el elemento "Luis" de la lista nombres
print(nombres[1])
```

```python
# Operaciones básicas con listas
mi_lista = [1, 2, 3, 4]

# Agregar un nuevo elemento
mi_lista.append(7)
print(mi_lista)
mi_lista.append(11)
print(mi_lista)

# Eliminar un elemento (final)
mi_lista.pop()
print(mi_lista)

# Agregar un elemento en una posición específica (insertar)
mi_lista.insert(4, 5)   # 4 => la posición, 5 => nuevo elemento
print(mi_lista)

# Ampliar o extender una lista
mi_lista.extend([10, 11, 12])
print(mi_lista)

# Eliminar un elemento por su valor
mi_lista.remove(7)
print(mi_lista)
```

```python
# Ejercicios con listas
notas = [10, 18, 7, 11, 5, 20, 13, 16]

# Hallar la cantidad de notas aprobatorias y desaprobatorias
aprobadas = 0
desaprobadas = 0

for n in notas:         # n => representa a un elemento de la lista
    if n > 10.5:        # Evaluamos si 'n' es mayor que 10.5 (aprobado)
        aprobadas += 1  # Acumulamos el contador
    else:
        desaprobadas += 1

print(f"Existen {aprobadas} notas aprobadas y {desaprobadas} desaprobadas")
```

```python
# Imprimir si un alumno es mujer (M) o varón (V) segun la siguiente lista de alumnos
alumnos = ["M", "M", "V", "M", "V", "M"]

# Recorremos la lista de alumnos
for a in alumnos:
    if a == "M":
        print("Es mujer")
    else:
        print("Es varón")
```

```python
# Contar cuantas muejeres y varones hay en la lista de alumnos
alumnos = ["M", "M", "V", "M", "V", "M"]

# Inicializamos los contadores
mujeres = 0
varones = 0

# Recorremos la lista de alumnos
for a in alumnos:
    if a == "M":
        mujeres += 1
    else:
        varones += 1

print(f"Existen {mujeres} mujeres y {varones} varones en la lista de alumnos")
```

```python
# Obtener el máximo y mínimo valor de una lista
notas = [10, 18, 7, 11, 5, 20, 13, 16, 12]

# Hallar la máxima y mínima nota
maxima = notas[0]   # Suponer que la nota máxima es el primer elemento
minima = notas[0]

for n in notas:
    if n > maxima:      # Comparar el elemento (iterado) con el valor actual de 'maxima'
        maxima = n      # Asignar el nuevo valor a 'maxima' si el elemento comparado es mayor
    if n < minima:
        minima = n

print(f"La máxima nota es: {maxima}")
print(f"La mínima nota es: {minima}")
```

```python
# Obtener el promedio de todas notas
suma = 0
for n in notas:
    suma += n

prom = suma / len(notas)
print(f"El promedio de notas es: {round(prom, 2)}")
```

```python
# Practica 1: Contar cuantos numeros pares hay en la lista de numeros
numeros = [10, 20, 30, 10, 15, 25, 35, 40, 50, 60, 13, 19, 22, 19, 44]

# Inicializamos el contador de números pares
pares = 0

# Recorremos la lista de números
for n in numeros:
    if n % 2 == 0:      # Evaluamos si el número es par
        pares += 1      # Acumulamos el contador de números pares

print(f"Existen {pares} números pares en la lista de números")
```

```python
# Practica 2: Calcular el monto total de ventas en una tienda según la siguiente lista de ventas.
# Además, calcular el promedio de ventas, la venta más alta y la venta más baja
ventas = [1000, 2000, 2770, 880, 1500, 3000, 2500, 980, 2210, 670]

# Inicializamos las variables para el total, máximo y mínimo
total_ventas = 0
max_venta = ventas[0]
min_venta = ventas[0]

# Recorremos la lista de ventas
for v in ventas:
    total_ventas += v      # Acumulamos el total de ventas
    if v > max_venta:      # Evaluamos si la venta actual es mayor que la venta máxima
        max_venta = v      # Asignamos el nuevo valor a la venta máxima
    if v < min_venta:      # Evaluamos si la venta actual es menor que la venta mínima
        min_venta = v      # Asignamos el nuevo valor a la venta mínima

# Calculamos el promedio de ventas
prom_ventas = total_ventas / len(ventas)

# Mostramos los resultados
print(f"El monto total de ventas es: {total_ventas}")
print(f"El promedio de ventas es: {prom_ventas:.2f}")
print(f"La venta más alta es: {max_venta}")
print(f"La venta más baja es: {min_venta}")

```

## Diccionarios

Un diccionario es una estructura de datos que se define entre llaves y está compuesto por elementos del tipo "key" y "value". Aqui un ejemplo:
<pre>
mi_diccionario = {
    "clave1": "valor1",
    "clave2": "valor2"
}
</pre>

* Las claves deben ser únicas e inmutables (por ejemplo: strings, números o tuplas).
* Los valores pueden ser de cualquier tipo (strings, números, listas, otros diccionarios, etc.).

### Ejemplos de diccinarios

#### ✅ Crear un diccionario
<pre>
persona = {
    "nombre": "Juan",
    "edad": 30,
    "ciudad": "Lima"
}
</pre>

#### ✅ Acceder a un valor
<pre>
print(persona["nombre"])  # Resultado: Juan
</pre>

#### ✅ Modificar un valor
<pre>
persona["edad"] = 31
print(persona["edad"])  # Resultado: 31
</pre>

#### ✅ Agregar un nuevo par clave-valor
<pre>
persona["profesion"] = "Ingeniero"
print(persona)
# {'nombre': 'Juan', 'edad': 31, 'ciudad': 'Lima', 'profesion': 'Ingeniero'}
</pre>

#### ✅ Eliminar una clave
<pre>
del persona["ciudad"]
print(persona)
# {'nombre': 'Juan', 'edad': 31, 'profesion': 'Ingeniero'}
</pre>

```python
# Trabajando con diccionarios
alumno1 = {"nombre": "Juan", "sexo": "M", "nota": 15}
alumno2 = {"nombre": "Luis", "sexo": "M", "nota": 12}
alumno3 = {"nombre": "Carmen", "sexo": "F", "nota": 16}

print("Nombre del alumno 1:", alumno1["nombre"])
print("Sexo del alumno 2:", alumno2["sexo"])    
print("Nota del alumno 3:", alumno3["nota"])
```

```python
# Recorrer un diccionario
alumno = {"nombre": "Juan", "sexo": "M", "nota": 15}
for clave in alumno:
    print(clave, ":", alumno[clave])

print ("---Recorriendo el diccionario con items:")
for clave, valor in alumno.items():
    print(clave, ":", valor)

print(alumno.items())  # Mostrar todos los pares clave-valor del diccionario
```

```python
# Lista de diccionarios
alumnos = [
            {"nombre": "Juan", "sexo": "M", "nota": 15},
            {"nombre": "Luis", "sexo": "M", "nota": 12},
            {"nombre": "Carmen", "sexo": "F", "nota": 16}
        ]

# Recorrer la lista de diccionarios
for alu in alumnos:
    print("Nombre:", alu["nombre"])
    print("Sexo:", alu["sexo"])
    print("Nota:", alu["nota"])
    print("---")  # Separador entre alumnos
```

```python
# Lista de trabajores
trabajadores = [
                {"nombre": "Ana", "edad": 30, "sueldo": 2500, "sexo": "F"},
                {"nombre": "Luis", "edad": 25, "sueldo": 2000, "sexo": "M"},
                {"nombre": "Carlos", "edad": 28, "sueldo": 2200, "sexo": "M"},
                {"nombre": "María", "edad": 32, "sueldo": 2800, "sexo": "F"},
                {"nombre": "Pedro", "edad": 27, "sueldo": 2100, "sexo": "M"}
            ]

# Contar la cantidad de trabajadores hombres y mujeres
hombres = 0
mujeres = 0

# Recorremos la lista de trabajadores
for tra in trabajadores:
    if tra["sexo"] == "M":
        hombres += 1
    else:
        mujeres += 1

print(f"Hombres: {hombres}, Mujeres: {mujeres}")
```

```python
# Lista de empleados
empleados = [
    {"nombre": "Ana", "edad": 30, "departamento": "Ventas", "sexo": "F"},
    {"nombre": "Luis", "edad": 25, "departamento": "Marketing", "sexo": "M"},
    {"nombre": "Pedro", "edad": 40, "departamento": "Recursos Humanos", "sexo": "M"},
    {"nombre": "María", "edad": 35, "departamento": "Finanzas", "sexo": "F"},
    {"nombre": "Laura", "edad": 28, "departamento": "IT", "sexo": "F"},
    {"nombre": "Carlos", "edad": 32, "departamento": "Ventas", "sexo": "M"}
]

# Contar empleados por sexo
hombres = 0
mujeres = 0

for emp in empleados:
    if emp["sexo"] == "M":
        hombres += 1
    if emp["sexo"] == "F":
        mujeres += 1

print(f"Existen {hombres} hombres y {mujeres} mujeres")
```

```python
# Con la lista anterior, encontrar al empleado más joven y más viejo
empleado_mas_joven = empleados[0]   # Suponer que el empleado más joven es el primero de la lista
empleado_mas_viejo = empleados[0]

# Recorremos la lista de empleados
for emp in empleados:
    if emp["edad"] < empleado_mas_joven["edad"]:    # Si la edad del empleado iterado es menor que del más joven
        empleado_mas_joven = emp
    if emp["edad"] > empleado_mas_viejo["edad"]:    # Si la edad del empleado iterado es mayor que del más viejo
        empleado_mas_viejo = emp


# Mostrar resultado
print(f"Empleado más joven: {empleado_mas_joven['nombre']}, Edad: {empleado_mas_joven['edad']}")
print(f"Empleado más viejo: {empleado_mas_viejo['nombre']}, Edad: {empleado_mas_viejo['edad']}")
```

```python
# Practica 1: El área de RR.HH. de una empresa registra la asistencia diaria de sus empleados:
empleados = [
    {"nombre": "Ana", "area": "Ventas", "hora_entrada": 8.10},
    {"nombre": "Luis", "area": "Sistemas", "hora_entrada": 9.15},
    {"nombre": "Carla", "area": "Contabilidad", "hora_entrada": 8.45},
    {"nombre": "Pedro", "area": "Ventas", "hora_entrada": 9.30},
    {"nombre": "María", "area": "Sistemas", "hora_entrada": 8.20}
]

# La empresa considera tardanza una entrada posterior a las 9.00 am. Se pide:
# - Mostrar el nombre y área de todos los empleados
# - Mostrar únicamente los empleados que llegaron tarde (Nombre y área)
# - Contar cuántos empleados llegaron tarde

# Recorremos la lista de empleados
print("Lista de empleados:")
print("-"*30)
for emp in empleados:
    print(f"Nombre: {emp['nombre']}, Área: {emp['area']}")

print("\nEmpleados que llegaron tarde:")
print("-"*30)
tardanza = 0
for emp in empleados:
    if emp["hora_entrada"] > 9.00:
        print(f"Nombre: {emp['nombre']}, Área: {emp['area']}")
        tardanza += 1

print(f"\nCantidad de empleados que llegaron tarde: {tardanza}")
```

```python
# Practica 2: Una tienda tiene el siguiente inventario de productos:
productos = [
    {"codigo": "P001", "nombre": "Laptop", "precio": 3500, "stock": 5},
    {"codigo": "P002", "nombre": "Mouse", "precio": 80, "stock": 25},
    {"codigo": "P003", "nombre": "Teclado", "precio": 150, "stock": 8},
    {"codigo": "P004", "nombre": "Monitor", "precio": 900, "stock": 3},
    {"codigo": "P005", "nombre": "Webcam", "precio": 250, "stock": 0}
]

# Se pide:
# - Mostrar los productos que tienen stock menor a 10 unidades (Código, Nombre, Precio y Stock)
# - Mostrar los productos agotados (Código, Nombre y Precio)
# - Calcular el valor total del inventario (Precio * Stock) y mostrarlo

# Recorremos la lista de productos
print("Productos con stock menor a 10 unidades:")
print("-"*50)
for prod in productos:
    if prod["stock"] < 10:
        print(f"Código: {prod['codigo']}, Nombre: {prod['nombre']}, Precio: {prod['precio']}, Stock: {prod['stock']}")

print("\nProductos agotados:")
print("-"*50)
for prod in productos:
    if prod["stock"] == 0:
        print(f"Código: {prod['codigo']}, Nombre: {prod['nombre']}, Precio: {prod['precio']}")

# Calcular el valor total del inventario
valor_total = 0
for prod in productos:
    valor_total += prod["precio"] * prod["stock"]

print(f"\nValor total del inventario: {valor_total}")
```
