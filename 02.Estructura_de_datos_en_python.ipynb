## Manejo de Listas en Python

- Las listas en Python, al igual que las cadenas, son datos ordenados. Una diferencia importante es que es un tipo de dato derivado, también llamado contenedor.
- Las listas contienen como elementos a enteros, a cadenas o incluso a otras listas. Una lista es una secuencia ordenada de datos.
- Las operaciones mas comunes son `append()` (agrega un elemento al final) y `pop()` (saca un elemento del final). 
- La longitud de una lista (o tamaño) está definida por la cantidad de elementos. Utilizamos la función `len()` para este propósito.
- Para acceder a los elementos de una lista lo hacemos mediante su índice que representa la posición del elemento empezando desde 0.

```python
# Listas de ejemplo
numeros = [10, 20, 30, 10]
nombres = ["Ana", "Luis", "Maria"]
mixta = ["Lima", 200, True, 125.88]

# Hallando la longitud de una lista
print(f"La longitud de la lista 'numeros' es: {len(numeros)}")

# Obtener el elemento "Luis" de la lista nombres
print(nombres[1])
```

```python
# Operaciones básicas con listas
mi_lista = [1, 2, 3, 4]

# Agregar un nuevo elemento
mi_lista.append(7)
print(mi_lista)
mi_lista.append(11)
print(mi_lista)

# Eliminar un elemento (final)
mi_lista.pop()
print(mi_lista)

# Agregar un elemento en una posición específica (insertar)
mi_lista.insert(4, 5)   # 4 => la posición, 5 => nuevo elemento
print(mi_lista)

# Ampliar o extender una lista
mi_lista.extend([10, 11, 12])
print(mi_lista)

# Eliminar un elemento por su valor
mi_lista.remove(7)
print(mi_lista)
```

```python
# Ejercicios con listas
notas = [10, 18, 7, 11, 5, 20, 13, 16]

# Hallar la cantidad de notas aprobatorias y desaprobatorias
aprobadas = 0
desaprobadas = 0
for n in notas:         # n => representa a un elemento de la lista
    if n > 10.5:        # Evaluamos si 'n' es mayor que 10.5 (aprobado)
        aprobadas += 1  # Acumulamos el contador
    else:
        desaprobadas += 1

print(f"Existen {aprobadas} notas aprobadas y {desaprobadas} desaprobadas")
```

```python
# Obtener el máximo y mínimo valor de una lista
notas = [10, 18, 7, 11, 5, 20, 13, 16, 12]

# Hallar la máxima y mínima nota
maxima = notas[0]   # Suponer que la nota máxima es el primer elemento
minima = notas[0]
for n in notas:
    if n > maxima:      # Comparar el elemento (iterado) con el valor actual de 'maxima'
        maxima = n      # Asignar el nuevo valor a 'maxima' si el elemento comparado es mayor
    if n < minima:
        minima = n

print(f"La máxima nota es: {maxima}")
print(f"La mínima nota es: {minima}")

# Obtener el promedio de todas notas
suma = 0
for n in notas:
    suma += n
prom = suma / len(notas)
print(f"El promedio de notas es: {round(prom, 2)}")
```

## Diccionarios

Un diccionario es una estructura de datos que se define entre llaves y está compuesto por elementos del tipo "key" y "value". Aqui un ejemplo:
<pre>
mi_diccionario = {
    "clave1": "valor1",
    "clave2": "valor2"
}
</pre>

* Las claves deben ser únicas e inmutables (por ejemplo: strings, números o tuplas).
* Los valores pueden ser de cualquier tipo (strings, números, listas, otros diccionarios, etc.).

### Ejemplos de diccinarios

#### ✅ Crear un diccionario
<pre>
persona = {
    "nombre": "Juan",
    "edad": 30,
    "ciudad": "Lima"
}
</pre>

#### ✅ Acceder a un valor
<pre>
print(persona["nombre"])  # Resultado: Juan
</pre>

#### ✅ Modificar un valor
<pre>
persona["edad"] = 31
print(persona["edad"])  # Resultado: 31
</pre>

#### ✅ Agregar un nuevo par clave-valor
<pre>
persona["profesion"] = "Ingeniero"
print(persona)
# {'nombre': 'Juan', 'edad': 31, 'ciudad': 'Lima', 'profesion': 'Ingeniero'}
</pre>

#### ✅ Eliminar una clave
<pre>
del persona["ciudad"]
print(persona)
# {'nombre': 'Juan', 'edad': 31, 'profesion': 'Ingeniero'}
</pre>

```python
# Trabajando con diccionarios
alumno1 = {"nombre": "Juan", "sexo": "M", "nota": 15}
alumno2 = {"nombre": "Luis", "sexo": "M", "nota": 12}
alumno3 = {"nombre": "Carmen", "sexo": "F", "nota": 16}

print("Nombre del alumno 1:", alumno1["nombre"])
print("Sexo del alumno 2:", alumno2["sexo"])    
print("Nota del alumno 3:", alumno3["nota"])
```

```python
# Recorrer un diccionario
alumno = {"nombre": "Juan", "sexo": "M", "nota": 15}
for clave in alumno:
    print(clave, ":", alumno[clave])

print ("---Recorriendo el diccionario con items:")
for clave, valor in alumno.items():
    print(clave, ":", valor)

print(alumno.items())  # Mostrar todos los pares clave-valor del diccionario
```

```python
# Lista de diccionarios
alumnos = [
            {"nombre": "Juan", "sexo": "M", "nota": 15},
            {"nombre": "Luis", "sexo": "M", "nota": 12},
            {"nombre": "Carmen", "sexo": "F", "nota": 16}
        ]

# Recorrer la lista de diccionarios
for alu in alumnos:
    print("Nombre:", alu["nombre"])
    print("Sexo:", alu["sexo"])
    print("Nota:", alu["nota"])
    print("---")  # Separador entre alumnos
```

```python
# Lista de empleados
empleados = [
    {"nombre": "Ana", "edad": 30, "departamento": "Ventas", "sexo": "F"},
    {"nombre": "Luis", "edad": 25, "departamento": "Marketing", "sexo": "M"},
    {"nombre": "Pedro", "edad": 40, "departamento": "Recursos Humanos", "sexo": "M"},
    {"nombre": "María", "edad": 35, "departamento": "Finanzas", "sexo": "F"},
    {"nombre": "Laura", "edad": 28, "departamento": "IT", "sexo": "F"},
    {"nombre": "Carlos", "edad": 32, "departamento": "Ventas", "sexo": "M"}
]

# Contar empleados por sexo
hombres = 0
mujeres = 0
for emp in empleados:
    if emp["sexo"] == "M":
        hombres += 1
    if emp["sexo"] == "F":
        mujeres += 1

print(f"Existen {hombres} hombres y {mujeres} mujeres")
```

```python
# Con la lista anterior, encontrar al empleado más joven y más viejo
empleado_mas_joven = empleados[0]   # Suponer que el empleado más joven es el primero de la lista
empleado_mas_viejo = empleados[0]
for emp in empleados:
    if emp["edad"] < empleado_mas_joven["edad"]:    # Si la edad del empleado iterado es menor que del más joven
        empleado_mas_joven = emp
    if emp["edad"] > empleado_mas_viejo["edad"]:    # Si la edad del empleado iterado es mayor que del más viejo
        empleado_mas_viejo = emp
# Mostrar resultado
print(f"Empleado más joven: {empleado_mas_joven['nombre']}, Edad: {empleado_mas_joven['edad']}")
print(f"Empleado más viejo: {empleado_mas_viejo['nombre']}, Edad: {empleado_mas_viejo['edad']}")
```
