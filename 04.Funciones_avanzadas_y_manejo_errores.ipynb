## 1. Expresiones lambda
Una expresión `lambda` es una forma corta de crear una función pequeña sin usar def. Por eso también se le suele llamar una función anónima.

La estructura básica es:
```python
lambda parametros: expresion
```

Por ejemplo, esta función tradicional:
```python
def sumar(a, b):
    return a + b
```

se puede escribir como lambda así:
```python
sumar = lambda a, b: a + b
```

````python
# Ejercicios 1: Crear una funcion lambda que devuelva el cuadrado de un numero
cuadrado = lambda num: num**2

# Mosrtrar el resultado de la funcion lambda
print(cuadrado(5))  # Output: 25
````

````python
# Ejercicio 2: Crear una funcion lambda que determina si una persona es mayor de edad (18 años o más)
es_mayor_de_edad = lambda edad: edad >= 18

# Mostrar el resultado de la funcion lambda
print(es_mayor_de_edad(20))  # Output: True
````

### ¿Cuándo usar def y cuándo lambda?
Usa `def` para funciones normales o cuando la lógica tiene varias instrucciones:
```python
def calcular_sueldo(sueldo):
    descuento = sueldo * 0.10
    sueldo_neto = sueldo - descuento
    return sueldo_neto
```

Usa `lambda` cuando la operación es corta y sencilla:
```python
calcular_igv = lambda monto: monto * 0.18
```

## 2. Usando map()
`map()` en Python se usa cuando quieres aplicar una misma función a cada elemento de una colección, por ejemplo, una lista, y obtener los resultados transformados.

La idea:
```bash
lista original
      ↓
   función
      ↓
map(función, lista)
      ↓
lista transformada
```

Su sintaxis básica es:
```python
map(funcion, iterable)
```

````python
# Ejercicio 1: Crear una funcion que duplique el valor de un numero y
# utilizarla para obtener el duplicado de los elementos de una lista
def duplicar(num):
    return num * 2

numeros = [1, 2, 3, 4, 5]
duplicados = list(map(duplicar, numeros))
print(duplicados)  # Output: [2, 4, 6, 8, 10]
````

````python
# Ejercicio 2: Realizar el ejercicio anterior utilizando una funcion lambda en lugar de una funcion definida
numeros = [1, 2, 3, 4, 5]
duplicados_lambda = list(map(lambda num: num * 2, numeros))
print(duplicados_lambda)  # Output: [2, 4, 6, 8, 10]
````
