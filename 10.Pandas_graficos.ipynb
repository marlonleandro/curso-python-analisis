## Introducción a Matplotlib

**Matplotlib** es una biblioteca de Python para crear gráficos de todo tipo. Su módulo principal, **pyplot**, es ideal para construir gráficos básicos rápidamente.

````python
# Importanto libreria pyplot de matplotlib
import matplotlib.pyplot as plt
````

## Gráficos básicos

### Gráfico de líneas

````python
# Graficando la evalución de ventas de los primeros 6 meses del año
meses = ["Ene", "Feb", "Mar", "Abr", "May", "Jun"]
ventas = [145, 180, 91, 200, 230, 265]

plt.plot(meses, ventas)
plt.title("Evolución de ventas")
plt.xlabel("Meses")  #Título del eje x
plt.ylabel("Ventas (miles de USD)")  #Título del eje y
plt.grid() #Incluye grillas en el gráfico
plt.show()
````

### Gráfico de barras

````python
# Comparar las ventas totales por tipo de producto
productos = ["Producto A", "Producto B", "Producto C"]
ventas = [450, 380, 570]
colores = ["blue", "green", "orange"]

plt.bar(productos, ventas, color=colores)
plt.title("Ventas totales por tipo de producto")
plt.xlabel("Producto")  #Título del eje x
plt.ylabel("Ventas (miles de USD)")  #Título del eje y
plt.show()
````

````python
# Gráfico de barras horizonal
plt.barh(productos, ventas, color="purple")
plt.title("Ventas totales por tipo de producto")
plt.xlabel("Ventas (miles de USD)")  #Título del eje x
plt.ylabel("Producto")  #Título del eje y
plt.show()
````

### Gráficos de dispersión (Scatter plot)

````python
# Visualizar la relación entre horas de estudio y las calificaciones
horas_estudio = [2, 3, 4, 5, 6, 7, 8, 9, 10]
calificaciones = [8, 10, 12, 15, 16, 17, 18, 19, 20]

plt.scatter(horas_estudio, calificaciones, color='red', marker='x')
plt.title("Relación entre horas de estudio y calificaciones")
plt.xlabel("Horas de estudio")  #Título del eje x
plt.ylabel("Calificación")  #Título del eje y
plt.show()
````

## Gráficos personalizados

````python
# Agregando estilo al gráfico anterior
# Visualizar la relación entre horas de estudio y las calificaciones
horas_estudio = [2, 3, 4, 5, 6, 7, 8, 9, 10]
calificaciones = [8, 10, 12, 15, 16, 17, 18, 19, 20]

plt.scatter(horas_estudio, calificaciones, color='red', marker='o')
plt.title("Relación entre horas de estudio y calificaciones", fontsize=14, fontweight="bold")
plt.xlabel("Horas de estudio", fontsize=12)  #Título del eje x
plt.ylabel("Calificación")  #Título del eje y
plt.grid()
plt.show()
````

````python
# Graficando la evalución de ventas de los primeros 6 meses del año
meses = ["Ene", "Feb", "Mar", "Abr", "May", "Jun"]
ventas = [145, 180, 91, 200, 230, 265]

plt.plot(meses, ventas, linestyle="--", color="red")
plt.title("Evolución de ventas", fontsize=12, fontweight="bold")
plt.xlabel("Meses", fontsize=10)  #Título del eje x
plt.ylabel("Ventas (miles de USD)", fontsize=10)  #Título del eje y
plt.axhline(y=200, color="blue", linestyle="-", label="Meta de ventas")
plt.grid() #Incluye grillas en el gráfico
plt.legend() # Agregamos la leyenda al gráfico
plt.show()
````

````python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
````

### Gráficos de correlación (heatmaps)

````python
especies = pd.read_csv("./data/surveys.csv")
especies.head()
````

````python
# Eliminando datos nulos (weight)
especiesna = especies.dropna()[["hindfoot_length","weight"]]
especiesna.head()
````

````python
# Correlacción entre longitud de pata y peso
especies_cor = especiesna.corr()
sns.heatmap(especies_cor, annot=True, cmap='coolwarm', linewidths=1)
````

````python
# Creando un grafico con seaborn con dataset de propinas
tips = sns.load_dataset("tips")
sns.histplot(data=tips, x="total_bill")
````

````python
# Creando gráfico de correlación con la data de propinas (propia de Seaborn)
correlacion_tips = tips.corr(numeric_only=True)
sns.heatmap(correlacion_tips)
plt.show()
````

## Ejemplos con data personalizada

````python
alumnos = pd.read_csv("./data/alumnos.csv", delimiter=",")
data_dpto = alumnos.groupby("Departamento")
conteo=data_dpto["DNI"].count()
conteo
````

````python
conteo.plot(kind="bar", title="Alumnos por departamento")
````

````python
# Genear grafico de pie por sexo
alu_sexo = alumnos.groupby("Sexo")
conteo = alu_sexo["DNI"].count()
conteo
````

````python
conteo.plot.pie(title="Alumnos por sexo", autopct="%0.1f %%")
````

````python
# Graficando data de ventas
sales = pd.read_csv("./data/sales.csv")

# Agrupando por typo de tienda
sales_type = sales.groupby("type")
sales_type_sum = sales_type["weekly_sales"].sum()
sales_type_sum.plot(kind="pie", autopct="%0.1f %%")
plt.title("Ventas totales por tipo de tienda")
plt.ylabel("")
plt.show()
````

````python
# Grafico de barras con data de censo USA 2017 de estados con población mayor a 10M
census = pd.read_csv("./data/acs2017_county_data.csv")
state_pop = census.groupby("State")
state_pop_sum = state_pop.agg({'TotalPop': 'sum'})
state_pop_sum_10m = state_pop_sum[state_pop_sum['TotalPop']>10000000]
state_pop_sum_10m.plot(kind="bar")

plt.title("Estados con población mayor a 10M", fontsize=12, fontweight='bold')
plt.xlabel("Estado", fontsize=11)  #Título del eje x
plt.ylabel("Pobblación", fontsize=11)  #Título del eje y
plt.show()
````

````python
# Ejercicio 1: Generar un gráfico de barras con la cantidad de alumnos de los departamentos ICA, LIMA, AREQUIPA, CUSCO
departamentos = ["ICA", "LIMA", "AREQUIPA", "CUSCO"]
alumnos_dpto = alumnos.groupby("Departamento")["DNI"].count()
alumnos_dpto = alumnos_dpto.loc[departamentos]
alumnos_dpto.plot(kind="bar", color="skyblue", title="Cantidad de alumnos por departamento")
````

````python
# Ejercicio 2: Crear un grafico de barras con la información del Censo de pero 2017 (archivo PERU_CENSO_2017.xlsx)
# mostrando a cantidad de habitantes mujeres de los departamentos ICA, LIMA, AREQUIPA, CUSCO
pobacion_peru = pd.read_excel("./data/PERU_CENSO_2017.xlsx", sheet_name="POB7")
pobacion_peru.head()
````

````python
# set_index para usar el departamento como índice y seleccionar la columna de mujeres
poblacion_mujeres = pobacion_peru[pobacion_peru['Departamento'].isin(["ICA", "LIMA", "AREQUIPA", "CUSCO"])]
poblacion_mujeres = poblacion_mujeres.set_index("Departamento")["Mujeres"]
poblacion_mujeres.plot(kind="bar", color="pink", title="Cantidad de mujeres por departamento")
````

````python
# set_index para usar el departamento como índice y seleccionar la columna de mujeres
poblacion_sexo = pobacion_peru[pobacion_peru['Departamento'].isin(["ICA", "LIMA", "AREQUIPA", "CUSCO"])]
poblacion_sexo = poblacion_sexo.set_index("Departamento")[["Hombres", "Mujeres"]]
poblacion_sexo.plot(kind="bar", color=["steelblue", "lightcoral"], title="Cantidad de habitantes por sexo y departamento")
````

````python
# Ejercicio 3: Crear un gráfico de pie con el porcentaje promedio que representa cada sexo en la población del estado de Alabama
# según el censo de la población de USA 2017
df_alabama_sexo = census[census['State'] == 'Alabama'][["Men", "Women"]]
df_alabama_sexo.head()
````

````python
sexo_prom = df_alabama_sexo.mean()
sexo_prom.plot.pie(autopct='%1.1f%%', startangle=90)
plt.title("Porcentaje promedio de hombres y mujeres en Alabama")
plt.show()
````
