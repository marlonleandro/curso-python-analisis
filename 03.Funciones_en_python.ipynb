## Funciones en Python

Las funciones en Python son bloques de código reutilizables que realizan una tarea específica. En lugar de escribir las mismas instrucciones varias veces, las colocamos dentro de una función y luego la ejecutamos cada vez que la necesitemos.

Por ejemplo, una máquina que recibe dos números y devuelve su suma:
```python
def sumar(a, b):
    resultado = a + b
    return resultado
```

````python
# Ejercicio 1: Saludo personalizado
def saludo(nombre):
    return f"¡Hola, {nombre}! Bienvenido/a."

# Llamar a la función con un nombre de ejemplo
nombre_usuario = "Juan"
print(saludo(nombre_usuario))
````

````python
# Ejercicio 2: Suma de dos números
def suma(a, b):
    return a + b

# Llamar a la función con dos números de ejemplo
num1 = 5
num2 = 3
print(f"La suma de {num1} y {num2} es: {suma(num1, num2)}")
````

````python
# Ejercicio 3: Calculadora de área de un rectángulo
def area_rectangulo(base, altura):
    return base * altura

# Llamar a la función con base y altura de ejemplo
base = 4
altura = 6
print(f"El área del rectángulo con base {base} y altura {altura} es: {area_rectangulo(base, altura)}")
````

````python
# Ejercicio 4: Verificar si una persona es mayor de edad
def es_mayor_de_edad(edad):
    return edad >= 18

# Llamar a la función con una edad de ejemplo
edad_usuario = 20
if es_mayor_de_edad(edad_usuario):
    print(f"Con {edad_usuario} años, eres mayor de edad.")
else:
    print(f"Con {edad_usuario} años, no eres mayor de edad.")
````

````python
# Ejercicio 5: Determinar si un número es positivo, negativo o cero
def determinar_signo(numero):
    if numero > 0:
        return "positivo"
    elif numero < 0:
        return "negativo"
    else:
        return "cero"

# Llamar a la función con un número de ejemplo
numero = -5
print(f"El número {numero} es {determinar_signo(numero)}.")
````

````python
# Ejercicio 6: Calcular el precio final de una compra
# Una tienda ofrece descuentos según el monto de compra:
# - Menos de S/ 100: sin descuento.
# - De S/ 100 a S/ 500: 5%.
# - Más de S/ 500: 10%.

def calcular_precio_final(monto_compra):
    if monto_compra < 100:
        descuento = 0
    elif 100 <= monto_compra <= 500:
        descuento = 0.05
    else:
        descuento = 0.10
    
    precio_final = monto_compra * (1 - descuento)
    return precio_final

# Llamar a la función con un monto de compra de ejemplo
monto = 250
precio_final = calcular_precio_final(monto)
print(f"El precio final de la compra de S/ {monto} es: S/ {precio_final:.2f}")
````

````python
# EJercicio 7: Calcular el promedio de una lista de números
def calcular_promedio(notas):
    if len(notas) == 0:
        return 0
    return sum(notas) / len(notas)

# Llamar a la función con una lista de notas de ejemplo
notas_estudiante = [18, 19, 10, 11, 13]
promedio = calcular_promedio(notas_estudiante)
print(f"El promedio de las notas {notas_estudiante} es: {promedio:.2f}")
````

````python
# Ejercicio 8: Encontrar el numero mayor de una lista de números
def encontrar_mayor(numeros):
    if len(numeros) == 0:
        return None
    return max(numeros)

# Llamar a la función con una lista de números de ejemplo
numeros = [3, 7, 2, 9, 5]
mayor_numero = encontrar_mayor(numeros)
if mayor_numero is not None:
    print(f"El número mayor de la lista {numeros} es: {mayor_numero}")
````

````python
# Ejercicio 9: Contar cuántas veces aparece un elemento en una lista
def contar_apariciones(lista, elemento):
    cantidad = 0
    for item in lista:
        if item == elemento:
            cantidad += 1
    return cantidad

# Llamar a la función con una lista y un elemento de ejemplo
lista_ejemplo = [1, 2, 3, 2, 4, 2, 5]
elemento_a_contar = 2
cantidad_apariciones = contar_apariciones(lista_ejemplo, elemento_a_contar)
print(f"El elemento {elemento_a_contar} aparece {cantidad_apariciones} veces en la lista.")

````

````python
# Ejercicio 10: Hallar la solución de una ecuación cuadrática ax^2 + bx + c = 0 usando la fórmula general.
def resolver_ecuacion_cuadratica(a, b, c):
    discriminante = b**2 - 4*a*c
    if discriminante < 0:
        return "No hay soluciones reales."
    elif discriminante == 0:
        x = -b / (2*a)
        return f"Una solución real: x = {x}"
    else:
        x1 = (-b + discriminante**0.5) / (2*a)
        x2 = (-b - discriminante**0.5) / (2*a)
        return f"Dos soluciones reales: x1 = {x1}, x2 = {x2}"

# Llamar a la función con coeficientes de ejemplo
a = 1
b = -3
c = 2
soluciones = resolver_ecuacion_cuadratica(a, b, c)
print(f"Las soluciones de la ecuación {a}x^2 + {b}x + {c} = 0 son: {soluciones}")
````
