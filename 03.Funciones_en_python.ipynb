## Funciones en Python

Las funciones en Python son bloques de código reutilizables que realizan una tarea específica. En lugar de escribir las mismas instrucciones varias veces, las colocamos dentro de una función y luego la ejecutamos cada vez que la necesitemos.

Por ejemplo, una máquina que recibe dos números y devuelve su suma:
```python
def sumar(a, b):
    resultado = a + b
    return resultado
```

````python
# Ejercicio 3: Calculadora de área de un rectángulo
def area_rectangulo(base, altura):
    return base * altura

# Llamar a la función con base y altura de ejemplo
base = 4
altura = 6
print(f"El área del rectángulo con base {base} y altura {altura} es: {area_rectangulo(base, altura)}")
````

````python
# Practica 1: Crear una funcion que permita calcular el area de un triangulo (usando la base y altura)
def area_triangulo(base, altura):
    area = (base * altura) / 2
    return area

# Llamar a la función con base y altura de ejemplo
base = 5
altura = 8
area = area_triangulo(base, altura)
print(f"El área del triángulo con base {base} y altura {altura} es: {area}")
````

````python
# Practica 2: Crear una funcion que calcule el area de una triangulo con sus tres lados (usando la formula de heron)
def area_triangulo_heron(a, b, c):
    s = (a + b + c) / 2  # Semiperímetro
    area = (s * (s - a) * (s - b) * (s - c)) ** 0.5  # Fórmula de Herón
    return area

# Llamar a la función con los tres lados de ejemplo
area1 = area_triangulo_heron(3, 4, 5)
area2 = area_triangulo_heron(5, 12, 13)
area3 = area_triangulo_heron(7, 24, 25)
print(f"El área del triángulo con lados 3, 4 y 5 es: {area1}")
print(f"El área del triángulo con lados 5, 12 y 13 es: {area2}")
print(f"El área del triángulo con lados 7, 24 y 25 es: {area3}")
````

````python
# Ejercicio 4: Verificar si una persona es mayor de edad
def es_mayor_de_edad(edad):
    return edad >= 18

# Llamar a la función con una edad de ejemplo
edad_usuario = 20
if es_mayor_de_edad(edad_usuario):
    print(f"Con {edad_usuario} años, eres mayor de edad.")
else:
    print(f"Con {edad_usuario} años, no eres mayor de edad.")
````

````python
# Practica 3: Crear una funciona que evalue si una persona es sujeto a credito
# - Evaluar por edad, ingresos, deuda y si tiene trabajo
def es_sujeto_a_credito(edad, ingresos, deuda, tiene_trabajo):
    if edad < 18:
        return False
    if ingresos < 1000:
        return False
    if deuda > 5000:
        return False
    if not tiene_trabajo:
        return False
    return True

# Llamar a la función con datos de ejemplo
edad_usuario = 25
ingresos_usuario = 2000
deuda_usuario = 3000
tiene_trabajo_usuario = False

tiene_credito = es_sujeto_a_credito(edad_usuario, ingresos_usuario, deuda_usuario, tiene_trabajo_usuario)
if tiene_credito:
    print("Eres sujeto a crédito.")
else:
    print("No eres sujeto a crédito.")
````

````python
# Ejercicio 5: Determinar si un número es positivo, negativo o cero
def determinar_signo(numero):
    if numero > 0:
        return "positivo"
    elif numero < 0:
        return "negativo"
    else:
        return "cero"

# Llamar a la función con un número de ejemplo
numero1 = -5
numero2 = 0
numero3 = 10

print(f"El número {numero1} es {determinar_signo(numero1)}.")
print(f"El número {numero2} es {determinar_signo(numero2)}.")
print(f"El número {numero3} es {determinar_signo(numero3)}.")
````

````python
# Ejercicio 6: Calcular el precio final de una compra
# Una tienda ofrece descuentos según el monto de compra:
# - Menos de S/ 100: sin descuento.
# - De S/ 100 a S/ 500: 5%.
# - Más de S/ 500: 10%.

def calcular_precio_final(monto_compra):
    if monto_compra < 100:
        descuento = 0
    elif 100 <= monto_compra <= 500:
        descuento = 0.05
    else:
        descuento = 0.10
    
    precio_final = monto_compra * (1 - descuento)
    return precio_final

# Llamar a la función con un monto de compra de ejemplo
monto1 = 250
precio_final1 = calcular_precio_final(monto1)
print(f"El precio final de la compra de S/ {monto1} es: S/ {precio_final1:.2f}")

monto2 = 600
precio_final2 = calcular_precio_final(monto2)
print(f"El precio final de la compra de S/ {monto2} es: S/ {precio_final2:.2f}")
````

````python
# EJercicio 7: Calcular el promedio de una lista de notas
def calcular_promedio(notas):
    if len(notas) == 0:
        return 0
    return sum(notas) / len(notas)

# Llamar a la función con una lista de notas de ejemplo
notas_estudiante = [18, 19, 10, 11, 13]
promedio = calcular_promedio(notas_estudiante)
print(f"El promedio de las notas {notas_estudiante} es: {promedio:.2f}")
````

````python
# Practica 4: Crear una funcion que devuelva una tupla con el maximo y minimo de una lista de notas
def max_min_notas(notas):
    if len(notas) == 0:     # Verificando si la lista está vacía
        return (None, None)
    max_nota = max(notas)
    min_nota = min(notas)
    return (max_nota, min_nota)

# Llamar a la función con una lista de notas de ejemplo
notas_estudiante = [18, 19, 10, 11, 13]
max_nota, min_nota = max_min_notas(notas_estudiante)

print(f"El máximo de las notas es: {max_nota}")
print(f"El mínimo de las notas es: {min_nota}")
````

````python
# Ejercicio 8: Encontrar el numero mayor de una lista de números
def encontrar_mayor(numeros):
    if len(numeros) == 0:
        return None
    return max(numeros)

# Llamar a la función con una lista de números de ejemplo
numeros = [3, 7, 2, 9, 5]
mayor_numero = encontrar_mayor(numeros)
if mayor_numero is not None:
    print(f"El número mayor de la lista {numeros} es: {mayor_numero}")

lista_vacia = []
mayor_numero_vacio = encontrar_mayor(lista_vacia)
if mayor_numero_vacio is None:
    print("La lista está vacía, no hay número mayor.")
````

````python
# Ejercicio 9: Contar cuántas veces aparece un elemento en una lista
def contar_apariciones(lista, elemento):
    cantidad = 0
    for item in lista:
        if item == elemento:
            cantidad += 1
    return cantidad

# Llamar a la función con una lista y un elemento de ejemplo
lista_ejemplo = [1, 2, 3, 2, 4, 2, 5]
elemento_a_contar = 2
cantidad_apariciones = contar_apariciones(lista_ejemplo, elemento_a_contar)
print(f"El elemento {elemento_a_contar} aparece {cantidad_apariciones} veces en la lista.")
````

````python
# Practica 5: Crear una funcion que devuelva un lista sin elementos repetidos
def eliminar_repetidos(lista):
    lista_sin_repetidos = []
    for item in lista:
        if item not in lista_sin_repetidos:
            lista_sin_repetidos.append(item)
    return lista_sin_repetidos

# Llamar a la función con una lista de ejemplo
lista_con_repetidos = [1, 2, 3, 2, 4, 3, 5, 1, 3, 5, 4]
lista_sin_repetidos = eliminar_repetidos(lista_con_repetidos)  
print(f"La lista sin elementos repetidos es: {lista_sin_repetidos}")
````

````python
# Practica 6: Crear una funcion que devuelva la suma de los n primeros numeros naturales
def suma_n_primeros_numeros(n):
    if n < 1:
        return 0
    sumatoria = n * (n + 1) / 2  # Usando la fórmula de la suma de los primeros n números naturales
    return sumatoria

# Llamar a la función con un número de ejemplo
num = 10
resultado = suma_n_primeros_numeros(num)
print(f"La suma de los {num} primeros números naturales es: {resultado}")
````

````python
# Ejercicio 10: Hallar la solución de una ecuación cuadrática ax^2 + bx + c = 0 usando la fórmula general.
def resolver_ecuacion_cuadratica(a, b, c):
    discriminante = b**2 - 4*a*c
    if discriminante < 0:
        return "No hay soluciones reales."
    elif discriminante == 0:
        x = -b / (2*a)
        return f"Una solución real: x = {x}"
    else:
        x1 = (-b + discriminante**0.5) / (2*a)
        x2 = (-b - discriminante**0.5) / (2*a)
        return f"Dos soluciones reales: x1 = {x1}, x2 = {x2}"

# Llamar a la función con coeficientes de ejemplo
a = 1
b = -3
c = 2
soluciones = resolver_ecuacion_cuadratica(a, b, c)
print(f"Las soluciones de la ecuación {a}x^2 + {b}x + {c} = 0 son: {soluciones}")
````

````python
# Practica 8: Realizar el mismo ejercicio de la ecuacion cuadratica pero devolver una tupla con las soluciones
def resolver_ecuacion_cuadratica_tupla(a, b, c):
    discriminante = b**2 - 4*a*c
    if discriminante < 0:
        return ()
    elif discriminante == 0:
        x = -b / (2*a)
        return (x,)
    else:
        x1 = (-b + discriminante**0.5) / (2*a)
        x2 = (-b - discriminante**0.5) / (2*a)
        return (x1, x2)

# Llamar a la función con coeficientes de ejemplo
a = 1
b = -3
c = 2
soluciones = resolver_ecuacion_cuadratica_tupla(a, b, c)
if len(soluciones) == 0:
    print(f"No hay soluciones reales para la ecuación.")
elif len(soluciones) == 1:
    print(f"Una solución real para la ecuación es: x = {soluciones[0]}")
else:
    print(f"Dos soluciones reales para la ecuación son: x1 = {soluciones[0]}, x2 = {soluciones[1]}")
````

````python
# Ejercicio 11: Calcular el sueldo total a pagar de una lista de empleados
# Cada empleado es un diccionario con su nombre y sueldo

def calcular_sueldo_total(empleados):
    total = 0   # Acumulador para el sueldo total
    for empleado in empleados:
        total += empleado["sueldo"]
    return total

# Llamar a la función con una lista de empleados
lista_empleados = [
    {"nombre": "Ana", "sueldo": 2000},
    {"nombre": "Luis", "sueldo": 1800},
    {"nombre": "Marta", "sueldo": 2200},
    {"nombre": "Pedro", "sueldo": 1700}
]

sueldo_total = calcular_sueldo_total(lista_empleados)
print(f"El sueldo total a pagar es: S/ {sueldo_total:.2f}")
````

````python
# Ejercicio 12: Crear 3 funciones que devuelvan la venta total, la cantidad de productos vendidos por tienda, el producto más vendido y el producto menos vendido.
# Cada transacción es un diccionario con el nombre de la tienda, el producto, la cantidad vendida y el precio unitario.

def calcular_venta_total(transacciones):
    total = 0
    for transaccion in transacciones:
        total += transaccion["cantidad"] * transaccion["precio_unitario"]
    return total

def calcular_cantidad_productos_vendidos_por_tienda(transacciones):
    cantidad_por_tienda = {}
    for transaccion in transacciones:
        tienda = transaccion["tienda"]
        cantidad = transaccion["cantidad"]
        if tienda in cantidad_por_tienda:
            cantidad_por_tienda[tienda] += cantidad
        else:
            cantidad_por_tienda[tienda] = cantidad
    return cantidad_por_tienda

def encontrar_producto_mas_y_menos_vendido(transacciones):
    ventas_por_producto = {}
    for transaccion in transacciones:
        producto = transaccion["producto"]
        cantidad = transaccion["cantidad"]
        if producto in ventas_por_producto:
            ventas_por_producto[producto] += cantidad
        else:
            ventas_por_producto[producto] = cantidad

    producto_mas_vendido = max(ventas_por_producto, key=ventas_por_producto.get)
    producto_menos_vendido = min(ventas_por_producto, key=ventas_por_producto.get)

    return producto_mas_vendido, producto_menos_vendido

# Llamar a las funciones con una lista de transacciones de ejemplo
transacciones = [
    {"tienda": "Tienda A", "producto": "Producto 1", "cantidad": 10, "precio_unitario": 5.5},
    {"tienda": "Tienda B", "producto": "Producto 2", "cantidad": 20, "precio_unitario": 3.8},
    {"tienda": "Tienda A", "producto": "Producto 1", "cantidad": 5, "precio_unitario": 5.5},
    {"tienda": "Tienda B", "producto": "Producto 3", "cantidad": 15, "precio_unitario": 4.3},
    {"tienda": "Tienda C", "producto": "Producto 2", "cantidad": 8, "precio_unitario": 3.8},
    {"tienda": "Tienda B", "producto": "Producto 1", "cantidad": 12, "precio_unitario": 5.5},
    {"tienda": "Tienda A", "producto": "Producto 3", "cantidad": 7, "precio_unitario": 4.3},
    {"tienda": "Tienda C", "producto": "Producto 2", "cantidad": 10, "precio_unitario": 3.8}
]

venta_total = calcular_venta_total(transacciones)
cantidad_por_tienda = calcular_cantidad_productos_vendidos_por_tienda(transacciones)
producto_mas_vendido, producto_menos_vendido = encontrar_producto_mas_y_menos_vendido(transacciones)

print(f"La venta total es: S/ {venta_total:.2f}")
print("Cantidad de productos vendidos por tienda:")
for tienda, cantidad in cantidad_por_tienda.items():
    print(f"{tienda}: {cantidad} productos vendidos")

print(f"El producto más vendido es: {producto_mas_vendido}")
print(f"El producto menos vendido es: {producto_menos_vendido}")
````
