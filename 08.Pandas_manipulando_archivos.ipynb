## Leyendo archivo CSV

```python
# Importamos la libreria
import pandas as pd

# Ubicamos la ruta del archivo
rutaArchivo = "data\\alumnos.csv"
df = pd.read_csv(rutaArchivo, delimiter=";")
df
```

```python
# Listar columnas
df.columns
```

## Agrupando datos

```python
# Agrupar datos por Departamento
df_dpto = df.groupby("Departamento")
df_dpto
```

```python
# Contar registros de la data agrupada
df_dpto["DNI"].count()
```

```python
# Agrupando por más de una columna
df_dpto_sex = df.groupby(["Departamento", "Sexo"])
df_dpto_sex["DNI"].count()
```

```python
# Mostrando solo un registro (indice)
df_dpto_sex["DNI"].count()["AREQUIPA"]
```

## Filtrando datos

```python
# Seleccionar alumnos del departamento de ICA
df["Departamento"] == "ICA"
```

```python
# Mostrando alumnos del departamento de ICA
df[df["Departamento"] == "ICA"]
```

```python
# Seleccionando algunas columnas
df_alu_ica = df[df["Departamento"] == "ICA"]
df_alu_ica[["Nombres", "Apellidos", "DNI", "Email"]]
```

```python
# Aplicando más de una condición
df[(df["Departamento"] == "ICA") & (df["Distrito"] == "PISCO")]
```

```python
# Buscando personas cuyo apellidos contenga la palabra Sanchez
df[df["Apellidos"].str.contains("sanchez", case=False)]
```

```python
# Buscando personas de departamento de ICA no sentitivo
df[df["Departamento"].str.fullmatch("ica", case=False)]
```

## Segmentando datos de un DataFrame

```python
# Leer archivos de alumnos
rutaArchivo = "data\\alumnos.csv"
df = pd.read_csv(rutaArchivo, delimiter=";")
df.head(10)
```

```python
# Filtrar datos por Hombres
df_hombres = df[df["Sexo"]=="M"]
df_hombres.head(10)
```

```python
# Filtrar datos por Mujeres
df_mujeres = df[df["Sexo"]=="F"]
df_mujeres.head(10)
```

```python
# Filtrar datos por Hombres del departamento de ICA - Método simple
df[(df["Sexo"] == "M") & (df["Departamento"] == "ICA")]
```

```python
# Filtrar datos por Hombres del departamento de ICA - Método con evaluación
df[df.eval("Sexo == 'M' & Departamento == 'ICA'")]
```
