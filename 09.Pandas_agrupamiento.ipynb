## Selección de datos

```python
# Importar la librería pandas para el análisis de datos
import pandas as pd
```

```python
sales = pd.read_csv("./data/sales.csv")
sales.head()
```

```python
# Seleccionando columnas específicas
sales1 = sales[["store", "type", "department", "date", "weekly_sales"]]
sales1.head()
```

```python
# Seleccioando todas las columnas a excepción de 2
# drop se usa para eliminar columnas del DataFrame (se liberan de la memoria)
columnas_excluidas = ["fuel_price_usd_per_l", "unemployment"]
sales2 = sales.drop(columns=columnas_excluidas)
sales2.head()
```

## Eliminando datos duplicados

```python
# Eliminar duplicados de la combinación store y type
# drop_duplicates se usa para eliminar filas duplicadas basadas en una combinación específica de columnas
stores_types = sales.drop_duplicates(subset=["store", "type"])
stores_types.head()
```

```python
# Eliminar duplicados de la combinación store y department
stores_depts = sales.drop_duplicates(subset=["store", "department"])
stores_depts.head()
```

```python
# Eliminar duplicados por fecha de las ventas en feriado
holiday_dates = sales[sales["is_holiday"]==True].drop_duplicates(subset="date")
holiday_dates.head()
```

## Conteo de datos

```python
# Conteo de tiendas por tipo
# value_counts se usa para contar la cantidad de ocurrencias de cada valor único en una columna
store_count = stores_types["type"].value_counts()
print(store_count)
```

```python
# Conteo de tiendas por tipo (representado en porcentaje)
store_prop = stores_types["type"].value_counts(normalize=True)
print(store_prop)
```

```python
# Counteo de tiendas por departamento ordenado de forma descendente
dept_counts = stores_depts["department"].value_counts(sort=True, ascending=False)
print(dept_counts)
```

## Agrupamiento de datos con groupby

```python
# Agrupando por typo de tienda
sales_type = sales.groupby("type")
sales_type["weekly_sales"].sum()
```

```python
# Calculando la venta total por el campo weekly_sales
sales_all = sales["weekly_sales"].sum()
print(sales_all)
```

```python
# Calculando la venta total de las tiendas de tipo A
sales_A = sales[sales["type"]=="A"]["weekly_sales"].sum()
print(sales_A)
```

```python
# Calculando la venta total de las tiendas de tipo B
sales_B = sales[sales["type"]=="B"]["weekly_sales"].sum()
print(sales_B)
```

```python
# Calculando la venta total de las tiendas de tipo C
sales_C = sales[sales["type"]=="C"]["weekly_sales"].sum()
print(sales_C)
```

```python
# Hallar la proporción de ventas semanales por tipo de tienda con respecto al total
sales_prop = [sales_A, sales_B, sales_C] / sales_all
print(sales_prop)
```

```python
# Hallando datos estadísticos con agrupación
sales_stats = sales.groupby("type")["weekly_sales"].agg(['min', 'max', 'sum', 'mean', 'median'])
print(sales_stats)
```

```python
# Hallando datos estadísticos con agrupación (de varias columnas)
sales_stats_multi = sales.groupby("type")[["weekly_sales", "fuel_price_usd_per_l"]].agg(['min', 'max'])
print(sales_stats_multi)
```

```python
# Agrupando dos columnas
sales_type_dept = sales.groupby(["type", "department"])["weekly_sales"].agg(['min', 'max', 'sum'])
print(sales_type_dept)
```

## Pivoteo de tablas

Pivotear es equivalente a transponer en Excel (pasar filas a columnas)

```python
# Ver la data original de ventas
sales.head()
```

```python
# Pivoteando la tabla con el promedio de las ventas semanales por tipo
# pivot_table se usa para reorganizar los datos y calcular agregados, en este caso el promedio de las ventas semanales por tipo
sales_by_type = sales.pivot_table(values="weekly_sales", index="type")
sales_by_type
```

```python
# Pivoteando la tabla de las ventas semanales por tipo, hallando max, min, sum y prom
# aggfunc se usa para especificar las funciones de agregación que se aplicarán a los datos, en este caso min, max, sum y mean
sales_by_type_stats = sales.pivot_table(values="weekly_sales", index="type", aggfunc=["min", "max", "sum", "mean"])
sales_by_type_stats
```

```python
# Pivoteando la tabla por dos columnas (departamento y tipo) con el total y promedio de las ventas semanales
sales_dept_type = sales.pivot_table(values="weekly_sales", index="department", columns="type", aggfunc=["sum", "mean"])
sales_dept_type
```

```python
# Pivoteando la tabla por dos columnas (departamento y tipo) con el total y promedio de las ventas semanales, llenado ceros en donde no haya data
sales_dept_type_z = sales.pivot_table(values="weekly_sales", index="department", columns="type", fill_value=0)
sales_dept_type_z
```

```python
# Pivoteando la tabla por dos columnas con el total y promedio de las ventas semanales, llenado ceros en donde no haya data y totalizada
sales_dept_type_zt = sales.pivot_table(values="weekly_sales", index="department", columns="type", fill_value=0, margins=True)
sales_dept_type_zt
```
